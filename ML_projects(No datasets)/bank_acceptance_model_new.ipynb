```python
import pandas, sklearn.preprocessing, sklearn.metrics, sklearn.model_selection, sklearn.svm, imblearn.under_sampling, pickle
```

```python
dataframe = pandas.read_csv("data/loan_approval_dataset.csv")
dataframe
```

```python
dataframe = dataframe.drop(["loan_id"], axis=1)
dataframe
```

```python
dataframe.info()
```

```python
dataframe.describe()
```

```python
dataframe[" loan_status"].value_counts()
```

```python
label_encoder = sklearn.preprocessing.LabelEncoder()
label_encoder
```

```python
scaler = sklearn.preprocessing.StandardScaler()
scaler
```

```python
under_sampler = imblearn.under_sampling.RandomUnderSampler(sampling_strategy="majority", random_state=42)
under_sampler
```

```python
for i in range(len(dataframe.columns)):
    if dataframe[dataframe.columns[i]].dtype == "str":
        dataframe[dataframe.columns[i]] = label_encoder.fit_transform(dataframe[dataframe.columns[i]])

dataframe
```

```python
X = dataframe[dataframe.columns[0:-1]]
Y = dataframe[dataframe.columns[-1]]
```

```python
X_train, X_test, Y_train, Y_test = sklearn.model_selection.train_test_split(X, Y, stratify=Y, test_size=0.3, random_state=42)
```

```python
X_train = pandas.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index)
X_test = pandas.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)
```

```python
X_train_resampled, Y_train_resampled = under_sampler.fit_resample(X_train, Y_train)
```

```python
Y_train_resampled.value_counts()
```

```python
model = sklearn.svm.SVC(kernel="rbf", random_state=42, gamma="scale")
model
```

```python
model.fit(X_train_resampled, Y_train_resampled)
```

```python
sklearn.metrics.accuracy_score(Y_test, model.predict(X_test))
```

```python
sklearn.metrics.precision_score(Y_test, model.predict(X_test))
```

```python
sklearn.metrics.recall_score(Y_test, model.predict(X_test))
```

```python
sklearn.metrics.f1_score(Y_test, model.predict(X_test))
```
